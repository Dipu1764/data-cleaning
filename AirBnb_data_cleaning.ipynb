```python
import pandas as pd

# Load the dataset
chicago_data = pd.read_csv('/content/listings chicago.csv')
new_orleans_data = pd.read_csv('/content/listings new Orleans.csv')

# Check the first few rows to understand the structure
print(chicago_data.head())
print(new_orleans_data.head())
```

```python
chicago_data.head()
```

```python
new_orleans_data.head()
```

```python
# Check for missing values
chicago_data.isnull().sum()
new_orleans_data.isnull().sum()
```

```python
chicago_data['city'] = 'Chicago'
new_orleans_data['city'] = 'New Orleans'

# Combine both datasets
combined_data = pd.concat([chicago_data, new_orleans_data])

# Save the combined data to a CSV file
combined_data.to_csv('airbnb_combined_data.csv', index=False)

# Verify by checking the first few rows
combined_data.head()
```
